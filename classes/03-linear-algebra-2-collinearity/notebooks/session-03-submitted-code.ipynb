```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Load Titanic training data from a public mirror.
url = "https://raw.githubusercontent.com/dsindy/kaggle-titanic/master/data/train.csv"
titanic = pd.read_csv(url)

# Select passenger fare and three predictors.
df = titanic[["Fare", "Pclass", "Age", "SibSp"]].copy()

# Fill missing ages with the median.
df["Age"] = df["Age"].fillna(df["Age"].median())

X = df[["Pclass", "Age", "SibSp"]]
y = df["Fare"]

# Fit the regression model.
model = LinearRegression()
model.fit(X, y)

print(df.head())
print("Intercept:", model.intercept_)
print("Coefficients:")
print(pd.Series(model.coef_, index=X.columns))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

features = ["Pclass", "Age", "SibSp"]
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for i, feature in enumerate(features):
    # vary one predictor
    values = np.linspace(df[feature].min(), df[feature].max(), 100)

    # keep the others at their medians
    inputs = pd.DataFrame({
        "Pclass": np.full(100, df["Pclass"].median()),
        "Age": np.full(100, df["Age"].median()),
        "SibSp": np.full(100, df["SibSp"].median())
    })
    inputs[feature] = values

    # predict fares
    predicted_fares = model.predict(inputs)

    # plot data and fitted line
    axes[i].scatter(
        df[feature], df["Fare"],
        alpha=0.25, s=15, label="observed fares"
    )
    axes[i].plot(
        values, predicted_fares,
        color="red", label="others fixed at medians"
    )
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel("fare (£)")
    axes[i].legend(fontsize=8)

plt.suptitle("conditional fitted lines")
plt.tight_layout()
plt.show()
```
